# How is the image?

```python
def howis(img):
  print('size = ',img.shape)
  print('max  = ',np.max(img))
  print('min  = ',np.min(img))
```

# Choose one image

```python
st_img = 'mandril.png'
#st_img = 'scale.png'
```

```python
img = imread(st_img)
X = img[:,:,0]
plt.figure(figsize=(8,8))
plt.imshow(X,cmap='gray')
plt.show()
```

# Spatial Sampling

```python
d = 8   # sampling each d pixels
(Nx,Mx) = X.shape
ix = range(0,Nx,d)
jx = range(0,Mx,d)
Ny = len(ix)
My = len(jx)
Y = np.zeros((Ny,My),np.uint8)
for i in range(Ny):
  for j in range(My):
    Y[i,j] = X[ix[i],jx[j]]
plt.figure(figsize=(8,8))
plt.imshow(Y,cmap='gray')
plt.show()
howis(Y)
```

# Histogram

```python
def imhist(X,n=256):
  (N,M) = X.shape
  h = np.zeros((n,))
  for i in range(N):
    for j in range(M):
      x = X[i,j]
      h[x] = h[x]+1
  return h
```

```python
n = 256
h = imhist(Y,n=n)
plt.figure(figsize=(12,8))
plt.plot(range(n),h[0:n])
plt.show()
```

# Quantization

```python
q = 8  # number of gray values
p = int(256/q)  # each p grayvalues
print(p)
(Nx,Mx) = X.shape
Ny = Nx
My = Mx
Y = np.zeros((Ny,My),np.uint8)
for i in range(Ny):
  for j in range(My):
    x = int(np.fix(X[i,j]/p)*p)
    Y[i,j] = x
plt.figure(figsize=(8,8))
plt.imshow(Y,cmap='gray')
plt.show()
howis(Y)
print('This image has = ',int(256/p),' grayvalues')
```

```python
n = 256
h = imhist(Y,n=n)
plt.figure(figsize=(12,8))
plt.plot(range(n),h[0:n])
plt.show()
```
